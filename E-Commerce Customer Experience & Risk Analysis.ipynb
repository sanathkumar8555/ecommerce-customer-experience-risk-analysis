```python
reviews_unique = (
    df_clean
    .dropna(subset=["review_id", "review_score"])
    .drop_duplicates(subset=["review_id"])
)
```

```python
total_reviews = reviews_unique["review_id"].count()
print("Total reviews:", total_reviews)
```

```python
one_star = (reviews_unique["review_score"] == 1).sum()
print("1-star reviews:", one_star)
```

```python
probability = one_star / total_reviews

print("Probability:", probability)
```

```python
percentage = probability * 100

print("1-star review percentage:", round(percentage, 2), "%")
```

```python
rating_count = reviews_unique["review_score"].value_counts().sort_index()
rating_probability = reviews_unique["review_score"].value_counts(
    normalize=True
).sort_index()
rating_percentage = rating_probability * 100
most_common_rating = reviews_unique["review_score"].mode()[0]

print("Most common rating:", most_common_rating)
```

```python
rating_distribution = pd.DataFrame({
    "Review Count": rating_count,
    "Probability": rating_probability,
    "Percentage": rating_percentage.round(2)
})

print(rating_distribution)
```

```python
import matplotlib.pyplot as plt

rating_count.plot(kind="bar")

plt.title("Customer Rating Distribution")
plt.xlabel("Review Score")
plt.ylabel("Number of Reviews")
plt.show()
```

```python
low_price = df_clean[
    df_clean["price"] <= 74.90
]["review_score"].dropna()

high_price = df_clean[
    df_clean["price"] > 74.90
]["review_score"].dropna()

print("Low-price average rating:", low_price.mean())
print("High-price average rating:", high_price.mean())
print("Low-price sample size:", len(low_price))
print("High-price sample size:", len(high_price))
```

```python
from scipy.stats import ttest_ind

order_data = (
    df_clean.groupby("order_id")
    .agg(
        avg_price=("price", "mean"),
        avg_rating=("review_score", "mean")
    )
    .dropna()
)

median_price = order_data["avg_price"].median()

low_group = order_data.loc[
    order_data["avg_price"] <= median_price, "avg_rating"
]

high_group = order_data.loc[
    order_data["avg_price"] > median_price, "avg_rating"
]

t_stat, p_value = ttest_ind(
    low_group,
    high_group,
    equal_var=False
)

print("Median price:", median_price)
print("Low-price average rating:", low_group.mean())
print("High-price average rating:", high_group.mean())
print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
from scipy.stats import ttest_ind

freight_data = (
    df_clean.groupby("order_id")
    .agg(
        avg_freight=("freight_value", "mean"),
        avg_rating=("review_score", "mean")
    )
    .dropna()
)


median_freight = freight_data["avg_freight"].median()

low_group = freight_data.loc[
    freight_data["avg_freight"] <= median_freight,
    "avg_rating"
]

high_group = freight_data.loc[
    freight_data["avg_freight"] > median_freight,
    "avg_rating"
]


t_stat, p_value = ttest_ind(
    low_group,
    high_group,
    equal_var=False
)


print("Median freight:", median_freight)
print("Low-freight average rating:", low_group.mean())
print("High-freight average rating:", high_group.mean())
print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
df_clean["price_category"] = pd.cut(
    df_clean["price"],
    bins=[0, 50, 150, float("inf")],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)
```

```python
price_analysis = (
    df_clean.dropna(subset=["review_score"])
    .groupby("price_category", observed=True)
    .agg(
        total_reviews=("review_score", "count"),
        negative_reviews=("is_low_rating", "sum")
    )
)

price_analysis["negative_rate"] = (
    price_analysis["negative_reviews"]
    / price_analysis["total_reviews"] * 100
)

print(price_analysis)
```

```python
plt.figure(figsize=(9, 5))

price_analysis["negative_rate"].plot(kind="bar")

plt.title("Negative Rating Percentage by Price Category")
plt.xlabel("Product Price Category")
plt.ylabel("Negative Rating (%)")

plt.xticks(rotation=0)
plt.show()
```

```python
monthly_reviews = (
    reviews_unique.copy()
)

monthly_reviews["review_month"] = pd.to_datetime(
    monthly_reviews["review_creation_date"]
).dt.to_period("M")

monthly_rating = (
    monthly_reviews.groupby("review_month")["review_score"]
    .mean()
)

print(monthly_rating)
```
